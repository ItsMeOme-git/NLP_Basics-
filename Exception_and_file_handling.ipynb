```python
with open("sample_text.txt", mode="w") as f:
  f.write("Python Programming is fun!")  # overwrites the existing content
```

```python
with open("sample_text.txt", mode="a") as f:
  f.write("Python Programming is cool!")  # append will only add new content to extsting content
```

```python
with open("sample_text.txt", mode="a") as f:
  f.write("\n 12345 \n 4567")  # append
```

```python
with open("sample_text.txt", mode="a") as f:
  f.writelines("4567890")  # appends as a new line due to writelines()
```

```python
with open("1.txt", "r") as f1:
  c1 = f1.read()
```

```python
with open("2.txt", "r") as f2:
  c2 = f2.read()
```

```python
with open("3.txt", "w") as f3:
  f3.writelines(c1)
  f3.writelines(c2)
```
